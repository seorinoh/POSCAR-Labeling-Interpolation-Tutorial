```python
# labeling POSCAR file parsing

# poscar_files들 중 poscar_labeling 파일을 지정 (0_Pmma_supercell.vasp를 poscar_labeling 파일로 함.)
poscar_labeling = poscar_files[0]
print(poscar_labeling)

# poscar 파일 내용을 줄 마다 정의하기 위해 읽기 모드로 열기
read_labeling_lines = open(poscar_labeling, 'r')
labeling_lines = read_labeling_lines.readlines()


# poscar 파일 줄별로 파싱하기

# 구조 이름
labeling_title = labeling_lines[0].strip()
print(labeling_title)


# 스케일 팩터
labeling_scale = float(labeling_lines[1].strip())
print(labeling_scale)


# float는 list 못받음 -> 문자열 하나만 받을 수 있음(즉, lattice 벡터 한 줄씩 실수화(float) 시킬 수 없음)
# 3x3 라티스 벡터(POSCAR 파일 2,3,4번째 줄)
# 한 줄씩, 한 줄당 3개의 좌표를 한 개씩 쪼개서 문자를 실수화(float) -> 한 줄 list만들고 -> 각 줄 합쳐서 list로 만들기
labeling_lattice = []

for i in range(2,5):
    lattice_row_list = labeling_lines[i].split()           #lattice_row_list = [a,b,c]의 리스트
    
    lattice_row = []

    for part in lattice_row_list:
        # a,b,c 좌표 하나씩 실수화 해서 [a,b,c] 리스트로 만들기
        lattice_row.append(float(part))

    #[a,b,c]/ [d,e,f]/ [g,h,i] 3개 리스트로 만들어서 전체 실수화 lattice 리스트 완성    
    labeling_lattice.append(lattice_row) 


# 각 원자의 종류
labeling_atom_types = labeling_lines[5].split()


# 각 원자 종류별 개수(공백 기준으로 원자 수 나눠서 각각 정수화 시켜 list로 만들기)
labeling_atom_counts = []
labeling_atom_counts_list = labeling_lines[6].split()

for part in labeling_atom_counts_list:
    labeling_atom_counts.append(int(part))
    
print(labeling_atom_counts)
print(type(labeling_atom_counts[0]))


# 원자들 총 개수
labeling_atoms_total = sum(labeling_atom_counts)
print(labeling_atoms_total)


# 원자 좌표의 형식
labeling_coord_type = labeling_lines[7].strip()
print(labeling_coord_type)


# 각 atom들 좌표
labeling_coords = []

for i in range(8,120):
    coords_row_list = labeling_lines[i].split() #coords_row_list = [a,b,c]의 리스트

    coords_row = []
    for part in coords_row_list:
        # a,b,c 좌표 하나씩 실수화 해서 [a,b,c] 리스트로 만들기
        coords_row.append(float(part)) 
    #[a,b,c]/ [d,e,f]/ [g,h,i] ...총 112개 리스트로 만들어서 전체 실수화 coord 리스트 완성 
    labeling_coords.append(coords_row) 
```

```python
# initial_POSCAR <-> labeling_POSCAR Atom Labeling 


# 루트 연산을 위한 math, numpy 모듈
from math import sqrt
import numpy as np


# 재정렬된 코드/ 측정된 거리 넣을 빈 리스트
label_reordered_coords=[]
measured_distance = []

# base_atom_num: 기준이 되는 initial 원자 번호(순서)
for base_atom_num in range(initial_atoms_total):

    row_measured_distance = []

    # target_atom_num: initial 하나의 원자 기준으로 비교하는 final 원자 번호(순서)
    for target_atom_num in range(labeling_atoms_total):

        new_distance_difference = []

        # initial과 final 원자의 좌표 한개씩 꺼내기
        for i_c, l_c in zip(initial_coords[base_atom_num], labeling_coords[target_atom_num]):
            # 원자 좌표 차이
            distance_difference = np.subtract (i_c, l_c)
            # 좌표 차이 반올림
            difference_round = np.round(distance_difference)
            # pbc 고려된 좌표 차이
            new_distance_difference.append(distance_difference - difference_round)

        # 행렬 곱셈하기 위해 벡터를 행렬로 바꾸기
        new_distance_difference = np.array(new_distance_difference)
        lattice = np.array(initial_lattice)
        # 실제 거리를 반영한 원자 좌표 차이
        real_distance = new_distance_difference @ lattice

        # 각 좌표 차이 제곱의 합
        total = real_distance[0]**2 + real_distance[1]**2 + real_distance[2]**2
        # 최종 루트 씌우기(거리 공식)
        distance = sqrt(total)
        row_measured_distance.append(distance)

    # 계산한 실제 거리 저장 
    measured_distance.append(row_measured_distance)

# a: initial 원자의 순서
for a in range (initial_atoms_total):
    # atom_distance: initial 원자 하나에 대한 final 원자 112개와의 거리 리스트
    atom_distance = measured_distance[a]
    # list 중 가장 작은 값 확인
    min_distance = min(atom_distance)
    # 가장 작은 값 인덱스 확인
    min_distance_index = atom_distance.index(min_distance)
    # 인덱스 값 정수화
    min_index = int(min_distance_index)
    # 거리 차이가 가장 작은 값의 코드를 reordered_coords에 정렬
    label_reordered_coords.append(labeling_coords[min_index])


# labeling해서 재정렬 된 파일 저장

# 파일 이름 설정하고 쓰기 모드로 열기
filename = f"0_Pmma_supercell_reordered.vasp"
with open(filename, 'w') as f:
      
    # 원자 이름 적기
    f.write(f"{labeling_title}\n")
    # 스케일 벡터 적기
    f.write(f"{labeling_scale}\n")
    # 라티스 벡터 한줄씩 적기
    for vec in labeling_lattice:
            f.write("  " + "  ".join(f"{x:.10f}" for x in vec) + "\n")
    # 원자 종류 적기
    f.write("  " + "  ".join(final_atom_types) + "\n")
    # 원자 종류별 갯수 적기
    f.write("  " + "  ".join(str(count) for count in labeling_atom_counts) + "\n")
    # 좌표 형식 적기
    f.write(f"{labeling_coord_type.capitalize()}\n")
    # 원자 좌표 한줄씩 적기
    for coord in label_reordered_coords:
        f.write("  " + "  ".join(f"{x:.10f}" for x in coord) + "\n")

    print("저장됨:", filename)
```

```python
# labeling_POSCAR <-> final_POSCAR Atom Labeling


# 루트 연산을 위한 math, numpy 모듈
from math import sqrt
import numpy as np


# 재정렬된 코드/ 측정된 거리 넣을 빈 리스트
reordered_coords=[]
measured_distance = []

# base_atom_num: 기준이 되는 initial 원자 번호(순서)
for base_atom_num in range(labeling_atoms_total):

    row_measured_distance = []

    # target_atom_num: initial 하나의 원자 기준으로 비교하는 final 원자 번호(순서)
    for target_atom_num in range(final_atoms_total):

        new_distance_difference = []

        # initial과 final 원자의 좌표 한개씩 꺼내기
        for l_c, f_c in zip(labeling_coords[base_atom_num], final_coords[target_atom_num]):
            # 원자 좌표 차이
            distance_difference = np.subtract (l_c, f_c)
            # 좌표 차이 반올림
            difference_round = np.round(distance_difference)
            # pbc 고려된 좌표 차이
            new_distance_difference.append(distance_difference - difference_round)

        # 행렬 곱셈하기 위해 벡터를 행렬로 바꾸기
        new_distance_difference = np.array(new_distance_difference)
        lattice = np.array(labeling_lattice)
        # 실제 거리를 반영한 원자 좌표 차이
        real_distance = new_distance_difference @ lattice

        # 각 좌표 차이 제곱의 합
        total = real_distance[0]**2 + real_distance[1]**2 + real_distance[2]**2
        # 최종 루트 씌우기(거리 공식)
        distance = sqrt(total)
        row_measured_distance.append(distance)

    # 계산한 실제 거리 저장 
    measured_distance.append(row_measured_distance)

# a: initial 원자의 순서
for a in range (labeling_atoms_total):
    # atom_distance: initial 원자 하나에 대한 final 원자 112개와의 거리 리스트
    atom_distance = measured_distance[a]
    # list 중 가장 작은 값 확인
    min_distance = min(atom_distance)
    # 가장 작은 값 인덱스 확인
    min_distance_index = atom_distance.index(min_distance)
    # 인덱스 값 정수화
    min_index = int(min_distance_index)
    # 거리 차이가 가장 작은 값의 코드를 reordered_coords에 정렬
    reordered_coords.append(final_coords[min_index])


print(reordered_coords)
print(len(reordered_coords))
print(reordered_coords[0])
print(reordered_coords[1])
print(reordered_coords[41])
print(reordered_coords[43])
```

```python
# (interpolation)선형 보간하는 함수 정의


# 선형 보간된 구조 lattice vector와 atom 좌표 
interpolated_structures = []

# n개의 중간 과정 만들기
n_steps = 15                            # 중간 과정 개수
for step in range(1, n_steps + 1):     #중간 과정 위치 (1부터 n_steps까지)
   

    # 보간 상수 x (중간 과정 위치 나타냄)
    # x는 곱해주는 비율(중간 과정을 3개로 만들면, 중간과정 1번 x=0.25/ 중간과정 2번 x=0.5/ 중간과정 3번 x=0.75)
    x = step / (n_steps + 1) 


    # lattice vector interpolation

    # 라티스 벡터 보간 후 중간 과정 벡터 저장할 빈 리스트
    intermediate_lattice = []

    # 3x3 라티스 벡터를 한 줄씩 지정
    for vec_initial, vec_final in zip(initial_lattice, final_lattice):

        # 좌표 하나씩 보간(interpolate)/ 보간 공식: interpol = ini + x * (fin - ini) 
        interpol_vec = [i_v + x * (f_v - i_v) for i_v, f_v in zip(vec_initial, vec_final)]

        # 중간 과정 lattice vector 저장
        intermediate_lattice.append(interpol_vec)
        

    # atom coords interpolation

    # 원자 좌표 보간 후 중간 과정 벡터 저장할 곳
    intermediate_coords = []

    # 원자 좌표를 한 줄씩 지정
    for row_initial, row_reordered in zip(initial_coords, reordered_coords):

        intermediate_coords_row =[]

        # 원자 좌표 한 개씩 지정
        for i_c, f_c in zip(row_initial, row_reordered):

            # pbc 고려해서 원자 좌표 차이 구하기
            interpol_difference = np.subtract (f_c, i_c)
            difference_round = np.round(interpol_difference)
            new_interpol_difference = interpol_difference - difference_round
        
            # 좌표 하나씩 보간(interpolate)/ 보간 공식: interpol = ini + x * (fin - ini)
            interpol_coords = i_c + x * (new_interpol_difference)

            # interpolated atom 1개 coords 저장
            intermediate_coords_row.append(interpol_coords)

        # 모든 interpolated atom의 coords 저장
        intermediate_coords.append(intermediate_coords_row)


    # 선형 보간된 구조 lattice vector와 atom 좌표로 저장
    interpolated_structures.append((intermediate_lattice, intermediate_coords))
    
print(interpolated_structures[0])
```

```python
# interpolation 파일들 저장 

# 파일 이름 설정하고 쓰기 모드로 열기
for i in range (n_steps):
    filename = f"POSCAR_interpolated_{i+1}.vasp"
    intermediate_lattice, intermediate_coords = interpolated_structures[i]
   
    with open(filename, 'w') as f:

        # 원자 이름 적기
        f.write(f"{initial_title}\n")

        # 스케일 벡터 적기
        f.write(f"{initial_scale}\n")

        # 라티스 벡터 한줄씩 적기
        for vec in intermediate_lattice:
                f.write("  " + "  ".join(f"{x:.10f}" for x in vec) + "\n")

        # 원자 종류 적기
        f.write("  " + "  ".join(initial_atom_types) + "\n")

        # 원자 종류별 갯수 적기
        f.write("  " + "  ".join(str(count) for count in initial_atom_counts) + "\n")

        # 좌표 형식 적기
        f.write(f"{initial_coord_type.capitalize()}\n")
        
        # 원자 좌표 한줄씩 적기
        for coord in intermediate_coords:
                f.write("  " + "  ".join(f"{x:.10f}" for x in coord) + "\n")

    print("저장됨:", filename)


# 저장된 'POSCAR_interpolated_1'파일 열어서 확인

filename = f"POSCAR_interpolated_{1}.vasp"
with open (filename, 'r') as f:
    content = f.read()

print(content)

      
```
